problem_text = """# Project Problem Statement: AI-Powered Financial Intelligence Engine

## 1. Objective
Build a point-in-time predictive classification engine for Indian equities to identify stocks that reach a 5% gain during the defined forward evaluation window.

## 2. Stock Universe
* **Segments:** Large-caps (Nifty 100) + Mid-caps (Nifty Midcap 150) + Filtered liquid small-caps (Average Daily Traded Value > ₹5 Crore).
* **Listing Age Rule:** Minimum 250 trading days required for listing age to ensure complete technical history.

## 3. Timeline & Windows
* **Feature Lookback Window:** 250 trading days prior to observation date T.
* **Observation Date (T):** All numerical, fundamental, and announcement features are strictly locked at or before date T (no look-ahead bias).
* **Forward Evaluation Window:** T+11 to T+15 (the 3rd trading week after observation point T).

## 4. Target Variable & Output
* **Ground Truth Label (y):** 
  * `1`: At least one closing price from T+11 through T+15 is at least 5% above the closing price on T.
  * `0`: No closing price from T+11 through T+15 reaches 5% above the closing price on T.
* **Model Output:** Probability score (P) via LightGBM to rank-order and filter high-conviction opportunities.

## 5. Initial Feature Blueprint
* **Direct Market Features:** OHLCV data, adjusted close, and corporate action flags.
* **Direct Fundamental Features:** P/E ratio, P/B ratio, ROE, and market capitalization from the latest available quarterly results.
* **Event/Announcement Features:** Structured flags for recent earnings releases, management changes, and corporate filings.
* **Derived Technical Features:** 
  * Short and medium-term historical returns (1m, 3m).
  * Moving average ratios (Close / 50-day SMA).
  * Rolling volatility (20-day standard deviation).
  * Volume surge ratio (Today's volume vs. 30-day average).
  * Relative strength vs. Nifty 50 over 30 days.
"""
